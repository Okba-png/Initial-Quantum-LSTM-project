```python
from qiskit import QuantumCircuit

# Prepare uniform superposition over 3 qubits
qc = QuantumCircuit(1)
qc.h([0])  # Apply h to the single qubit
print(qc.draw())
# This creates (|0> + |1>) / sqrt(2)
```

```python
from qiskit import QuantumCircuit

# Prepare uniform superposition over 3 qubits
qc = QuantumCircuit(2) # Create a quantum circuit with 2 qubits
qc.h([0,1])  # Apply H to all qubits simultaneously
print(qc.draw())
# This creates (|00> + |01> + |10> + |11>) / sqrt(4)
```

```python
from qiskit import QuantumCircuit

# Prepare uniform superposition over 3 qubits
qc = QuantumCircuit(3)
qc.h([0,1,2])  # Apply H to all qubits simultaneously
print(qc.draw())
# This creates (|000> + |001> + |010> + |011> + |100> + |101> + |110> + |111>) / sqrt(8)
```

```python
from qiskit import QuantumCircuit

# Erstelle einen QuantumCircuit mit 2 Qubits
# KEINE klassischen Bits → wir messen NICHT
qc1 = QuantumCircuit(2)

# Hadamard-Gate auf Qubit 0 
# Superposition erzeugen: |0> → (|0> + |1>) / sqrt(2)
qc1.h(0)

# CNOT: Qubit 0 kontrolliert Qubit 1
# Das erzeugt den Bell-State:
# (|00> + |11>) / sqrt(2)
qc1.cx(0, 1)


# Circuit anzeigen

print(qc1.draw())
```

```python
# Schritt 1: Qiskit importieren
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

# Schritt 2: Circuit mit 2 Qubits erstellen
# WICHTIG: KEINE klassischen Bits, KEINE Messung
qc2 = QuantumCircuit(2)

# Schritt 3: Superposition erzeugen
qc2.h(0)  # Hadamard auf Qubit 0

# Schritt 4: Entanglement erzeugen
qc2.cx(0, 1)  # CNOT: Qubit 0 kontrolliert Qubit 1

# Schritt 5: Circuit anzeigen (vor save_statevector, damit das Diagramm sauber bleibt)
print(qc2.draw())

# Schritt 6: Statevector im Circuit speichern (DAS FEHLTE)
qc2.save_statevector()

# Schritt 7: Statevector-Simulator auswählen
sim = AerSimulator(method="statevector")

# Schritt 8: Circuit für den Simulator kompilieren
compiled = transpile(qc2, sim)

# Schritt 9: Simulation ausführen
result = sim.run(compiled).result()

# Schritt 10: Statevector holen
state = result.get_statevector()

# Schritt 11: Statevector anzeigen
print("\nStatevector:")
print(state)
```
