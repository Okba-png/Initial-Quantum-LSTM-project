## 1. Theorie in Kürze

**Definition (globale Phase).** Die Zustände $|\psi\rangle$ und $e^{i\gamma}|\psi\rangle$ sind physikalisch **identisch**, denn alle Wahrscheinlichkeiten $|\langle x|\psi\rangle|^2$ bleiben gleich. Nur die **relative** Phase zwischen $|0\rangle$ und $|1\rangle$ zählt.

**Definition (Bloch-Kugel).** Jeder reine Ein-Qubit-Zustand lässt sich bis auf globale Phase schreiben als

$$|\psi\rangle=\cos\tfrac{\theta}{2}|0\rangle+e^{i\varphi}\sin\tfrac{\theta}{2}|1\rangle,\qquad \theta\in[0,\pi],\ \varphi\in[0,2\pi)$$

Der zugehörige **Bloch-Vektor** $\vec r=(\langle X\rangle,\langle Y\rangle,\langle Z\rangle)=(\sin\theta\cos\varphi,\ \sin\theta\sin\varphi,\ \cos\theta)$ ist ein Punkt auf der Einheitskugel.

**Gatter im Überblick:**

| Gatter | Matrix | Wirkung auf der Bloch-Kugel | Qiskit |
|---|---|---|---|
| $X$ | $\begin{pmatrix}0&1\\1&0\end{pmatrix}$ | $\pi$-Drehung um $x$-Achse | `qc.x(i)` |
| $Y$ | $\begin{pmatrix}0&-i\\i&0\end{pmatrix}$ | $\pi$-Drehung um $y$-Achse | `qc.y(i)` |
| $Z$ | $\begin{pmatrix}1&0\\0&-1\end{pmatrix}$ | $\pi$-Drehung um $z$-Achse | `qc.z(i)` |
| $S$ | $\begin{pmatrix}1&0\\0&i\end{pmatrix}$ | $\pi/2$-Drehung um $z$ | `qc.s(i)` |
| $T$ | $\begin{pmatrix}1&0\\0&e^{i\pi/4}\end{pmatrix}$ | $\pi/4$-Drehung um $z$ | `qc.t(i)` |
| $R_x(\vartheta)$ | $e^{-i\vartheta X/2}$ | beliebige Drehung um $x$ | `qc.rx(theta, i)` |
| $R_y(\vartheta)$ | $e^{-i\vartheta Y/2}$ | beliebige Drehung um $y$ | `qc.ry(theta, i)` |
| $R_z(\vartheta)$ | $e^{-i\vartheta Z/2}$ | beliebige Drehung um $z$ | `qc.rz(theta, i)` |

Die Rotation $R_y(\vartheta)|0\rangle=\cos\frac\vartheta2|0\rangle+\sin\frac\vartheta2|1\rangle$ ist besonders nützlich, weil sie die Wahrscheinlichkeit $p(1)=\sin^2\frac\vartheta2$ direkt einstellt.

**Messbasen.** Qiskit misst immer in der $Z$-Basis. Für eine Messung in der $X$-Basis wendet man vorher $H$ an, für die $Y$-Basis vorher $S^\dagger$ und dann $H$.

## 2. Werkzeuge für die Übungen

Diese Zelle einmal ausführen. Sie lädt alle Importe und definiert zwei Hilfsfunktionen.

```python
# Importe: Circuit, exakter Zustand, Operatoren, Pauli-Matrizen, Simulator
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector, Operator, Pauli
from qiskit_aer import AerSimulator

def bloch_vektor(state):
    # Bloch-Vektor (<X>, <Y>, <Z>) eines 1-Qubit-Zustands über Erwartungswerte
    return tuple(round(float(state.expectation_value(Pauli(p)).real), 4) for p in "XYZ")

def messen(qc, shots=1024):
    # Führt einen Circuit mit Messung auf dem Aer-Simulator aus und gibt die Häufigkeiten zurück
    sim = AerSimulator()
    compiled = transpile(qc, sim)
    return sim.run(compiled, shots=shots).result().get_counts()
```

### Demo: Bloch-Vektor und Bloch-Kugel

```python
# Zustand |+> = H|0> erzeugen
qc = QuantumCircuit(1)
qc.h(0)
zustand = Statevector(qc)

# Erwartungswerte: |+> liegt auf der positiven x-Achse -> (1, 0, 0)
print("Bloch-Vektor von |+>:", bloch_vektor(zustand))

# Grafische Darstellung auf der Bloch-Kugel (benötigt matplotlib)
from qiskit.visualization import plot_bloch_multivector
plot_bloch_multivector(zustand)
```

## 4. Lösungen

### Lösung 1
$Y|0\rangle=\begin{pmatrix}0&-i\\i&0\end{pmatrix}\begin{pmatrix}1\\0\end{pmatrix}=\begin{pmatrix}0\\i\end{pmatrix}=i|1\rangle$.
Die Wahrscheinlichkeit für `1` ist $|i|^2=1$. Der Faktor $i$ ist eine **globale Phase** und ohne physikalische Wirkung. Das Ergebnis ist also physikalisch dasselbe wie bei $X$.

```python
qa = QuantumCircuit(1)
qa.y(0)                                   # Y-Gatter auf |0>
za = Statevector(qa)
print(za)                                 # Amplitude 1j auf |1>
print({str(k): round(float(v), 4) for k, v in za.probabilities_dict().items()})  # Wahrscheinlichkeit für '1' ist 1
```

### Lösung 2
Von Hand: $HZH=\frac12\begin{pmatrix}1&1\\1&-1\end{pmatrix}\begin{pmatrix}1&0\\0&-1\end{pmatrix}\begin{pmatrix}1&1\\1&-1\end{pmatrix}=\frac12\begin{pmatrix}1&-1\\1&1\end{pmatrix}\begin{pmatrix}1&1\\1&-1\end{pmatrix}=\frac12\begin{pmatrix}0&2\\2&0\end{pmatrix}=X$.
Anschaulich: $H$ vertauscht die $x$- und $z$-Achse der Bloch-Kugel, deshalb wird aus einer $Z$-Drehung eine $X$-Drehung.

```python
# Circuit für H, Z, H (Reihenfolge: zuerst H, dann Z, dann H)
q_hzh = QuantumCircuit(1)
q_hzh.h(0)
q_hzh.z(0)
q_hzh.h(0)

# Vergleichs-Circuit: nur X
q_x = QuantumCircuit(1)
q_x.x(0)

# Matrizen ausgeben und Gleichheit prüfen
print(np.round(Operator(q_hzh).data, 4))
print("HZH == X :", Operator(q_hzh).equiv(Operator(q_x)))
```

### Lösung 3
Aus $p(1)=\sin^2\frac\vartheta2=0{,}75$ folgt $\sin\frac\vartheta2=\frac{\sqrt3}{2}$, also $\frac\vartheta2=\frac\pi3$ und $\vartheta=\frac{2\pi}{3}\approx2{,}094$.

```python
theta = 2 * np.arcsin(np.sqrt(0.75))       # analytisch: theta = 2*arcsin(sqrt(p))
print("theta =", round(theta, 4), "=", round(theta / np.pi, 4), "* pi")

# Exakte Prüfung mit dem Statevector
q3 = QuantumCircuit(1)
q3.ry(theta, 0)
print({str(k): round(float(v), 4) for k, v in Statevector(q3).probabilities_dict().items()})

# Prüfung mit Messungen (Zufallsschwankung erwartet, ca. 3072 mal '1')
q3.measure_all()
print(messen(q3, shots=4096))
```

### Lösung 4
(a) $S^2=\mathrm{diag}(1,i^2)=\mathrm{diag}(1,-1)=Z$. (b) $T^2=\mathrm{diag}(1,e^{i\pi/2})=S$.
(c) $R_x(\pi)=\cos\frac\pi2\,\mathbb 1-i\sin\frac\pi2\,X=-iX$. Die globale Phase ist also $-i$.

```python
# (a) S*S == Z
c_a = QuantumCircuit(1); c_a.s(0); c_a.s(0)
c_z = QuantumCircuit(1); c_z.z(0)
print("(a) S*S == Z :", Operator(c_a).equiv(Operator(c_z)))

# (b) T*T == S
c_b = QuantumCircuit(1); c_b.t(0); c_b.t(0)
c_s = QuantumCircuit(1); c_s.s(0)
print("(b) T*T == S :", Operator(c_b).equiv(Operator(c_s)))

# (c) Rx(pi) gleich X bis auf globale Phase
c_c = QuantumCircuit(1); c_c.rx(np.pi, 0)
c_x = QuantumCircuit(1); c_x.x(0)
print("(c) Rx(pi) ~ X :", Operator(c_c).equiv(Operator(c_x)))
print("Matrix von Rx(pi):\n", np.round(Operator(c_c).data, 4))   # entspricht -i * X
```

### Lösung 5
Vor der Messung in der $X$-Basis wendet man $H$ an: $H|+\rangle=|0\rangle$ und $H|-\rangle=|1\rangle$. Erwartung: für $|+\rangle$ immer `0`, für $|-\rangle$ immer `1`. Die vorher unsichtbare **Vorzeichen-Information** (Phase) wird so messbar.

```python
# |+> vorbereiten, Basiswechsel mit H, dann messen
q5a = QuantumCircuit(1)
q5a.h(0)          # Zustand |+>
q5a.h(0)          # Basiswechsel: X-Basis -> Z-Basis
q5a.measure_all()
print("|+> in X-Basis:", messen(q5a))     # erwartet: nur '0'

# |-> vorbereiten (H, dann Z), Basiswechsel, messen
q5b = QuantumCircuit(1)
q5b.h(0)
q5b.z(0)          # Zustand |->
q5b.h(0)          # Basiswechsel
q5b.measure_all()
print("|-> in X-Basis:", messen(q5b))     # erwartet: nur '1'
```

### Lösung 7
Aufbau: $H$ auf Qubit 0, dann zwei CNOTs von Qubit 0 auf Qubit 1 und 2. Erlaubt sind nur `000` und `111`, jeweils etwa 50 %.
$$|000\rangle\xrightarrow{H}\tfrac{1}{\sqrt2}(|000\rangle+|001\rangle)\xrightarrow{\mathrm{CX}(0,1)}\tfrac{1}{\sqrt2}(|000\rangle+|011\rangle)\xrightarrow{\mathrm{CX}(0,2)}\tfrac{1}{\sqrt2}(|000\rangle+|111\rangle)$$

```python
# Circuit mit 3 Qubits
ghz = QuantumCircuit(3)
ghz.h(0)          # Superposition auf Qubit 0
ghz.cx(0, 1)      # Qubit 1 folgt Qubit 0
ghz.cx(0, 2)      # Qubit 2 folgt Qubit 0

print(ghz.draw())
print(Statevector(ghz))                     # Amplituden nur bei Index 0 ('000') und 7 ('111')

# Messung: nur '000' und '111' dürfen auftreten
ghz_m = ghz.copy()
ghz_m.measure_all()
print(messen(ghz_m))
```

## 5. Zusammenfassung

- Globale Phasen sind unsichtbar, relative Phasen nicht.
- Jedes Ein-Qubit-Gatter ist eine Drehung auf der Bloch-Kugel.
- $R_y(\vartheta)$ stellt Wahrscheinlichkeiten gezielt ein: $p(1)=\sin^2\frac\vartheta2$.
- Messung in anderen Basen gelingt durch einen Basiswechsel vor der $Z$-Messung.
- Verschränkung lässt sich auf mehrere Qubits erweitern (GHZ).
